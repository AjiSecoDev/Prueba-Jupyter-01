## 1. Fórmulas (LaTeX)

Fórmula en línea: $E = mc^2$ y la famosa ecuación de Black-Scholes:

$$
C(S, t) = S \cdot N(d_1) - K e^{-r(T-t)} \cdot N(d_2)
$$

donde

$$
d_1 = \frac{\ln(S/K) + (r + \sigma^2/2)(T-t)}{\sigma \sqrt{T-t}}, \qquad d_2 = d_1 - \sigma \sqrt{T-t}
$$

## 7. Regresión lineal

Ajustamos un modelo de regresión lineal $y = a \cdot x + b$ a datos simulados con `numpy`:

```python
np.random.seed(7)
x = np.linspace(0, 10, 50)
y_true = 2.5 * x + 4.0
y = y_true + np.random.normal(0, 1.5, x.size)

a, b = np.polyfit(x, y, 1)
y_pred = a * x + b

ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - ss_res / ss_tot

print(f'Pendiente (a) = {a:.4f}')
print(f'Intercepto (b) = {b:.4f}')
print(f'R^2 = {r2:.4f}')
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(x, y, s=25, alpha=0.7, label='Datos observados')
ax.plot(x, y_pred, 'r-', linewidth=2, label=f'Regresión: y = {a:.2f}x + {b:.2f}')
ax.plot(x, y_true, 'g--', linewidth=1.5, label='Valor real (sin ruido)')
ax.set_title(f'Regresión lineal (R² = {r2:.4f})')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
plt.show()
```

```python
pred_df = pd.DataFrame({'x': [2.0, 5.0, 8.0, 10.0]})
pred_df['y_predicho'] = a * pred_df['x'] + b
pred_df
```

## 1. Fórmulas (LaTeX)

Fórmula en línea: $E = mc^2$ y la famosa ecuación de Black-Scholes:

$$
C(S, t) = S \cdot N(d_1) - K e^{-r(T-t)} \cdot N(d_2)
$$

donde

$$
d_1 = \frac{\ln(S/K) + (r + \sigma^2/2)(T-t)}{\sigma \sqrt{T-t}}, \qquad d_2 = d_1 - \sigma \sqrt{T-t}
$$

```python
np.random.seed(42)
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x) + np.random.normal(0, 0.1, x.size)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), 'b-', label='sin(x)')
ax.scatter(x, y, s=8, alpha=0.6, label='datos con ruido')
ax.set_title('Gráfico generado en el notebook')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
plt.show()
```
